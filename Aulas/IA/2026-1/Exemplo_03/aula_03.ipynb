```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)
n_players = 500

vel = np.random.randint(50, 99, n_players)
chute = np.random.randint(40, 99, n_players)
drible = np.random.randint(50, 99, n_players)
```

```python
valor_real = (vel * 2000) + (chute * 1500) + (drible * 1000) - 50000 + np.random.randint(-5000, 5000, n_players)

df = pd.DataFrame({'Velocidade': vel, 'Chute': chute, 'Drible': drible, 'Valor em Real': valor_real})
print(df)
```

```python
x = df[['Velocidade', 'Chute', 'Drible']]
y = df['Valor em Real']

model = LinearRegression()
model.fit(x, y)
```

```python
novo_jogador = np.array([[50, 40, 50]])
valor_previsto = model.predict(novo_jogador)

print(f"Status do jogador: 95 Velocidade, 70 chute, 88 Drible")
print(f"Valor estimado pela IA: {valor_previsto[0]:,.2f}")
print(f"Velocidade: {model.coef_[0]:.1f}, Chute: {model.coef_[1]:.1f}, Drible: {model.coef_[2]:.1f}")
```
